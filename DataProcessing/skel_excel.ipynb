```python
import os
import numpy as np
import pandas as pd

folder_path = r"/home/nyam/Ms_Thesis/Autism/Project_v3/DataProcessing/v2/skel_clean"  # Change this to your folder path
output_excel = 'skeleton_frames.xlsx'

all_rows = []
row_id = 1

for filename in os.listdir(folder_path):
    if filename.endswith('.npy'):
        # Parse class, subject_id, and clip_id from the filename
        parts = filename.split('_')
        class_name = parts[1]
        subject_id = f"{parts[1]}_{parts[2]}"
        clip_id = f"{parts[1]}_{parts[2]}_{parts[-1].replace('.npy', '')}"

        # Load the npy file: shape (frames, 33, 3)
        filepath = os.path.join(folder_path, filename)
        data = np.load(filepath)

        # Loop through each frame
        for frame in data:
            joint_coords = frame.flatten().tolist()  # 33*3 = 99 values
            row = [row_id, class_name, subject_id, clip_id] + joint_coords
            all_rows.append(row)
            row_id += 1

# Define column names
columns = ['s.no', 'class', 'subject_id', 'clip_id']
for j in range(33):
    columns += [f'joint_{j}_x', f'joint_{j}_y', f'joint_{j}_z']

# Create DataFrame and save to Excel
df = pd.DataFrame(all_rows, columns=columns)
df.to_excel(output_excel, index=False)

print(f"Saved to {output_excel}")
```
