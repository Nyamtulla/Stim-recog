```python
def slice_video_segments(video_path, output_folder, segment_title):
    try:
        with VideoFileClip(video_path) as video:
            fps = video.fps  # Frames per second
            total_frames = int(video.duration * fps)
            frame_count = 0
            segment_index = 0

            while frame_count < total_frames:
                start_time = frame_count / fps
                end_frame_count = min(frame_count + 64, total_frames)
                end_time = end_frame_count / fps

                subclip = video.subclip(start_time, end_time)

                # Check if the subclip has more than 30 frames
                if end_frame_count - frame_count >= 64:
                    # Save the segment
                    output_filename = os.path.join(output_folder, f"{segment_title}_slice_{segment_index}.mp4")
                    subclip.write_videofile(output_filename, codec='libx264')
                    segment_index += 1

                frame_count += 64  # Move to the next set of frames

        return True
    except Exception as e:
        print(f"Error slicing video '{segment_title}': {e}")
        return False


input_folder = 'action_segs'  # Folder where your cut video segments are stored
sliced_folder = 'sliced_clips'
os.makedirs(sliced_folder, exist_ok=True)

# Preprocess and slice videos
for video_file in os.listdir(input_folder):
    if video_file.endswith('.mp4'):  # Filter only for mp4 files
        video_path = os.path.join(input_folder, video_file)
        video_title = os.path.splitext(video_file)[0]  # Remove file extension from title
        slice_video_segments(video_path, sliced_folder, video_title)

print("slicing complete.")
```
